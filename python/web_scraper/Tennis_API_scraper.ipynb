```python
import os
import json
import requests
```

```python
HOST = 'tennis-api-atp-wta-itf.p.rapidapi.com'
BASE_URL = f"https://{HOST}/tennis/v2"

API_KEY = "a998f31b2dmsh051ce4c1f040c6dp1980f2jsn1f5abdbc5989"

HEADERS = {
    "X-RapidAPI-Key": API_KEY,
    "X-RapidAPI-Host": HOST,
}
```

```python
request_count = 0

def get(path: str, params: dict | None = None) -> dict:
    global request_count
    request_count += 1
    url = f"{BASE_URL}{path}"
    resp = requests.get(url, headers=HEADERS, params=params, timeout=15)
    print(f"[{request_count}] GET {resp.url} -> {resp.status_code}")
    resp.raise_for_status()
    return resp.json()

def show(data, limit=1500):
    text = json.dumps(data, indent=2, ensure_ascii=False)
    print(text[:limit] + ("...\n<truncated>" if len(text) > limit else ""))
```
